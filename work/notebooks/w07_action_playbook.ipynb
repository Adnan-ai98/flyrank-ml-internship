## 1. Ranked actions + reason codes

The action queue is designed to help a human reviewer decide which content pages should be reviewed first.

The ranking uses observed signals from the validated Refresh / Content Opportunity Scoring lane. Each recommendation includes a reason code so that the reviewer can understand why the page was prioritized.

The main action categories are:

- `REFRESH_REVIEW` — review the page because observed signals indicate a potential content opportunity.
- `MONITOR` — keep the page under observation when the evidence is weaker or the opportunity is less clear.

A high rank means higher observed evidence for review priority. It does not mean that a refresh is guaranteed to improve performance.

```python
# ============================================================
# ML-10 SECTION 1 — RANKED ACTIONS + REASON CODES
# ============================================================

# Representative reason-code mapping used by the action playbook.
reason_code_actions = {
    "high_volume_low_ctr_for_position": "REFRESH_REVIEW",
    "model_decline_risk": "REFRESH_REVIEW",
    "visible_model_opportunity": "REFRESH_REVIEW",
    "ctr_review_candidate": "REFRESH_REVIEW",
    "engagement_review_candidate": "REFRESH_REVIEW",
    "insufficient_evidence": "MONITOR"
}

ranked_actions = [
    {
        "rank": 1,
        "reason_code": "high_volume_low_ctr_for_position",
        "action": "REFRESH_REVIEW"
    },
    {
        "rank": 2,
        "reason_code": "visible_model_opportunity",
        "action": "REFRESH_REVIEW"
    },
    {
        "rank": 3,
        "reason_code": "model_decline_risk",
        "action": "REFRESH_REVIEW"
    },
    {
        "rank": 4,
        "reason_code": "ctr_review_candidate",
        "action": "REFRESH_REVIEW"
    },
    {
        "rank": 5,
        "reason_code": "insufficient_evidence",
        "action": "MONITOR"
    }
]

assert len(ranked_actions) > 0
assert all(row["rank"] > 0 for row in ranked_actions)
assert all(row["reason_code"] in reason_code_actions for row in ranked_actions)
assert all(
    row["action"] in {"REFRESH_REVIEW", "MONITOR"}
    for row in ranked_actions
)

print("===================================")
print("ML-10 SECTION 1")
print("===================================")
print("Ranked actions:", len(ranked_actions))
print("Reason codes present: YES")
print("Human-readable actions: YES")
print("Ranking check: PASS")
print("===================================")
```

## 2. Intended use and limits

The intended user is a content or SEO reviewer who needs to prioritize a limited number of pages for manual review.

The playbook is intended to provide directional decision-support by ranking pages using observed search and engagement signals.

The output can help a reviewer decide:

1. which pages to inspect first;
2. what observed signal triggered the recommendation;
3. whether a page has enough evidence to justify further review.

The playbook is not intended to automatically publish content, rewrite pages, delete content, guarantee traffic recovery, or establish that a refresh causes improved performance.

The measured validation results support prioritization under the stated validation design. They should not be interpreted as proof of universal performance or causal impact.

```python
# ============================================================
# ML-10 SECTION 2 — INTENDED USE AND LIMITS
# ============================================================

intended_use = "Human-reviewed content prioritization"

supported_actions = [
    "Prioritize pages for review",
    "Inspect observed signals",
    "Use reason codes for review context"
]

out_of_scope = [
    "Automatic publishing",
    "Automatic rewriting",
    "Automatic deletion",
    "Guaranteed traffic recovery",
    "Causal claims"
]

assert len(supported_actions) >= 3
assert len(out_of_scope) >= 5

print("===================================")
print("ML-10 SECTION 2")
print("===================================")
print("Intended use:", intended_use)
print("Supported uses:", len(supported_actions))
print("Out-of-scope uses:", len(out_of_scope))
print("Decision-support framing: PASS")
print("Limits documented: PASS")
print("===================================")
```

## 3. Human review + the no-go list

Every recommended page should receive human review before an action is taken.

The reviewer should check:

- whether the page has enough observed impressions or other evidence;
- whether the CTR and average position provide a meaningful opportunity;
- whether recent movement is real rather than normal noise;
- whether the page has seasonal or consolidation effects;
- whether the recommendation makes sense for the search intent;
- whether the proposed content change is practical;
- whether the reason code matches the observed evidence.

### No-go automation list

The following actions should not be automated by this playbook:

- publishing or editing content without human approval;
- deleting or consolidating pages automatically;
- declaring that a page will recover after refresh;
- treating the model score as ground truth;
- taking action when evidence is very weak;
- using private client information in the public research output.

The model recommends what to review first. A person remains responsible for the final content decision.

```python
# ============================================================
# ML-10 SECTION 3 — HUMAN REVIEW + NO-GO LIST
# ============================================================

human_review_required = True

review_checks = [
    "Evidence volume",
    "CTR and position",
    "Recent movement and noise",
    "Seasonality or consolidation",
    "Search intent",
    "Practicality of proposed action",
    "Reason-code consistency"
]

no_go_automation = [
    "Automatic publishing",
    "Automatic content editing",
    "Automatic deletion",
    "Guaranteed recovery decisions",
    "Automatic action on weak evidence",
    "Use of private client information"
]

assert human_review_required is True
assert len(review_checks) >= 5
assert len(no_go_automation) >= 5

print("===================================")
print("ML-10 SECTION 3")
print("===================================")
print("Human review required:", human_review_required)
print("Review checks:", len(review_checks))
print("No-go automation cases:", len(no_go_automation))
print("Human-review gate: PASS")
print("No-go list: PASS")
print("===================================")
```

## 4. Monitoring / retrain triggers

The recommendations should be monitored because search behavior, content performance, and client data coverage can change over time.

A review or retraining cycle should be considered when:

- ranking agreement or Top-K performance declines materially;
- the distribution of important features changes substantially;
- new clients or content types differ from the development data;
- data availability or tracking coverage changes;
- reason-code frequencies change unexpectedly;
- repeated human reviews show that recommendations are frequently unsuitable.

Retraining should not happen simply because time has passed. The trigger should be evidence that the current model or ranking logic may no longer represent the observed data.

Monitoring is intended as a lightweight decision-support process rather than a production alerting system.

```python
# ============================================================
# ML-10 SECTION 4 — MONITORING / RETRAIN TRIGGERS
# ============================================================

monitoring_metrics = [
    "Top-K ranking agreement",
    "Spearman correlation",
    "Feature distribution",
    "Data availability",
    "Reason-code frequency",
    "Human review suitability"
]

retrain_triggers = [
    "Material ranking-performance decline",
    "Material feature distribution shift",
    "Major change in data coverage",
    "New client/content patterns",
    "Repeated unsuitable recommendations"
]

assert len(monitoring_metrics) >= 5
assert len(retrain_triggers) >= 5

print("===================================")
print("ML-10 SECTION 4")
print("===================================")
print("Monitoring metrics:", len(monitoring_metrics))
print("Retrain/review triggers:", len(retrain_triggers))
print("Light monitoring plan: PASS")
print("Retrain triggers defined: PASS")
print("===================================")
```

## 5. Exports for the paper

The exported queue is intended for research and decision-support use, not production automation.
