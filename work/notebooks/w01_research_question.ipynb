## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*Question: Does average search position show a useful directional relationship with CTR that can help identify pages worth reviewing? The decision is which pages should receive further SEO investigation. An SEO/content team could use the finding to prioritize pages for review rather than treating every page equally. A wrong recommendation could waste time and resources on pages that do not need attention, so the result should be treated as decision-support rather than a guaranteed recommendation.

```python
print("Initial correlation:",
      round(df["avg_position"].corr(df["ctr"]), 3))

print("This is an observational check; it does not establish causation.")
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 pages. The observed correlation between avg_position and ctr is -0.073, which indicates a weak negative relationship. This makes the question worth investigating further, but the relationship is not strong enough to treat position as a standalone explanation for CTR.

```python
print("Number of pages:", len(df))
print("Average CTR:", round(df["ctr"].mean(), 4))
print("avg_position ↔ ctr correlation:",
      round(df["avg_position"].corr(df["ctr"]), 3))
```

## 4. Careful words: what I can and can't claim

I can report an observed or directional relationship between average position and CTR in this dataset. I can use the result as decision-support for deciding which pages may deserve further investigation. I cannot claim that average position causes CTR changes, that the relationship applies to every search situation, or that I have predicted or proved Google's algorithm. Any future recommendation will be presented as an observation from the available data.

```python
corr = df["avg_position"].corr(df["ctr"])

print(f"Observed correlation: {corr:.3f}")
print("Interpretation: weak directional relationship.")
print("No causal or Google-algorithm claim is being made.")
```
